```python
import sys          
import time         
import math         
from functools import lru_cache  

# алгоритм вычисления суммы, с помощью рекурсии
def recursive_sum(x, i):
    if i == 0:
        return 0
    current = x
    for _ in range(i):
        current = math.sin(current)
    return current + recursive_sum(x, i-1)

# 2 итеративная функция
def iterative_sum(x, n):
    total = 0
    current = x
    for _ in range(n):
        current = math.sin(current)
        total += current
    return total

# кеширование
def recursive_memoized(x, i, memo=None):
    if memo is None:
        memo = {}
    if i == 0:
        return 0
    if (x, i) in memo:
        return memo[(x, i)]
    current = x
    for _ in range(i):
        current = math.sin(current)
    result = current + recursive_memoized(x, i-1, memo)
    memo[(x, i)] = result
    return result


# с декоратором
@lru_cache(maxsize=None)  
def recursive_lru(x, i):
    if i == 0:
        return 0
    current = x
    for _ in range(i):
        current = math.sin(current)
    return current + recursive_lru(x, i-1)

# текстирование производительности
def performance_test():
    test_cases = [(0.5, 5), (0.5, 10), (0.5, 50), (0.5, 100)]
    
    for x, n in test_cases:
        print(f"\nТест: x={x}, n={n}")
        
        
        start = time.time()          
        r1 = recursive_sum(x, n)     
        t1 = time.time() - start     
        print(f"Рекурсивная: {t1:.6f} сек")
        
        
        start = time.time()
        r2 = iterative_sum(x, n)
        t2 = time.time() - start
        print(f"Итеративная: {t2:.6f} сек")
        
        
        start = time.time()
        r3 = recursive_memoized(x, n)
        t3 = time.time() - start
        print(f"С мемоизацией: {t3:.6f} сек")
        

        
        start = time.time()
        r5 = recursive_lru(x, n)
        t5 = time.time() - start
        print(f"LRU декоратор: {t5:.6f} сек")


        
        print(f"Результаты: {r1:.10f}, {r2:.10f}")
        print(f"Совпадают: {abs(r1 - r2) < 1e-10}")

# оцениваем верхнюю границу размерности задачи, для которой при рекурсивной реализации не происходит переполнение стека вызовов
def find_max_recursion_depth():
    x = 0.5
    max_depth = 0
    

    for depth in range(100, 10000, 100):
        try:
            recursive_sum(x, depth)
            
            max_depth = depth
        except RecursionError:
            # Если произошла ошибка, прерываем цикл
            break
    
    # Возвращаем максимальную глубину
    return max_depth


if __name__ == "__main__":
    x = 0.5
    n = 5
    
    
    print("Рекурсивный результат:", recursive_sum(x, n))
    print("Итеративный результат:", iterative_sum(x, n))
    print("С мемоизацией:", recursive_memoized(x, n))
    print("С LRU декоратором:", recursive_lru(x, n))
    
    performance_test()
    
    max_depth = find_max_recursion_depth()
    print(f"\nМаксимальная глубина рекурсии: {max_depth}")
    
    # Описываем блок-схемы алгоритмов
    print("\nБлок-схемы алгоритмов:")
    print("1. Рекурсивный: Начало -> Если i=0: вернуть 0 -> Вычислить sin^i(x) -> Прибавить рекурсивный вызов для i-1 -> Конец")
    print("2. Итеративный: Начало -> total=0, current=x -> Цикл n раз: current=sin(current), total+=current -> Вернуть total -> Конец")
    print("3. С мемоизацией: Начало -> Проверить кэш -> Если нет в кэше: вычислить, сохранить -> Использовать кэш -> Конец")
Рекурсивный результат: 2.233672287969179
Итеративный результат: 2.233672287969179
С мемоизацией: 2.233672287969179
Безопасный стек: 2.233672287969179
С LRU декоратором: 2.233672287969179

Тест: x=0.5, n=5
Рекурсивная: 0.000002 сек
Итеративная: 0.000001 сек
С мемоизацией: 0.000003 сек
Безопасная стек: 0.000002 сек
LRU декоратор: 0.000001 сек
Результаты: 2.2336722880, 2.2336722880
Совпадают: True

Тест: x=0.5, n=10
Рекурсивная: 0.000005 сек
Итеративная: 0.000001 сек
С мемоизацией: 0.000007 сек
Безопасная стек: 0.000002 сек
LRU декоратор: 0.000005 сек
Результаты: 4.1584627069, 4.1584627069
Совпадают: True

Тест: x=0.5, n=50
Рекурсивная: 0.000063 сек
Итеративная: 0.000004 сек
С мемоизацией: 0.000073 сек
Безопасная стек: 0.000005 сек
LRU декоратор: 0.000068 сек
Результаты: 15.0144328374, 15.0144328374
Совпадают: True

Тест: x=0.5, n=100
Рекурсивная: 0.000251 сек
Итеративная: 0.000006 сек
С мемоизацией: 0.000252 сек
Безопасная стек: 0.000010 сек
LRU декоратор: 0.000184 сек
Результаты: 24.3052944217, 24.3052944217
Совпадают: True
 
```
